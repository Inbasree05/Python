```python
# Create an array `[10, 25, 30, 45, 50, 60]`. Print only the elements greater than 30.
a=np.array([10,25,30,45,50,60])
print(a[a>30])
```

```python
# Create an array `[10, 20, 30, 40, 50]`. Print elements greater than 10 AND less than 50.
a=np.array([10,20,30,40,50])
print(a[(a>10) & (a<50)])
```

```python
# Create an array `[50, 10, 40, 20, 30]`. Sort it in ascending and descending order.
a=np.array([50,10,40,20,30])
print(np.sort(a))
print(np.sort(a)[::-1])
```

```python
# Create a 2D array `[[30, 10, 20], [60, 40, 50]]`. Sort each row and then sort each column.
a=np.array([[30,10,20],[60,40,50]])
print(np.sort(a,axis=1))
print(np.sort(a,axis=0))
```

```python
# Q1. Copy

# Create an array [10, 20, 30, 40]. Make a copy named b. Change the second element of b to 200. Print both arrays.
a=np.array([10,20,30,40])
b=a.copy()
b[1]=200
print(a)
print(b)
```

```python
# Q2. View

# Create an array [5, 10, 15, 20]. Make a view named b. Change the third element of b to 150. Print both arrays.
a=np.array([5,10,15,20])
b=a.view()
b[2]=150
print(a)
print(b)
```

```python
# Q3. Predict the output
a = np.array([1, 2, 3])

b = a.copy()
c = a.view()

b[0] = 100
c[1] = 200

print(a)
print(b)
print(c)
```

```python
# 1)Create an array [10, 20, 30, 40, 50] and print each element using a for loop.
a=np.array([10,20,30,40,50])
for x in a:
    print(x)
```

```python
# Q2. 2D iteration
a=np.array([[1,2,3],[4,5,6]])
for x in a:
    print(x)
```

```python
# Q4. np.nditer()
a=np.array([[10, 20],[30, 40]])
for x in np.nditer(a):
    print(x)
```

```python
# Q5. np.ndenumerate()
a=np.array([[10, 20],[30, 40]])
for index,value in np.ndenumerate(a):
    print(index,value)
```
